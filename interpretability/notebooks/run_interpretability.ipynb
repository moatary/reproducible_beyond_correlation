## 1. PCA eigenimages (baseline)
PCA components are globally orthogonal, producing coarse, spatially blurred eigenimages.


```python
from interpretability.extract_eigenimages import extract_weights, save_eigenimage_panel
W_pca = extract_weights("PCA", "mnist", n_components=16, seed=42)
save_eigenimage_panel(W_pca, "figures/PCA_mnist_eigenimages.png",
                      title="PCA Eigenimages (MNIST)")
from IPython.display import Image; Image("figures/PCA_mnist_eigenimages.png")
```

## 2. WDIWCD eigenimages (proposed)
The independence criterion produces sharper, more locally specific patterns.

![WDIWCD eigenimages](../figures/WDIWCD_mnist_eigenimages.png)


```python
W_wdiwcd = extract_weights("WDIWCD", "mnist", n_components=16, seed=42)
save_eigenimage_panel(W_wdiwcd, "figures/WDIWCD_mnist_eigenimages.png",
                      title="WDIWCD Eigenimages (MNIST)")
Image("figures/WDIWCD_mnist_eigenimages.png")
```

## 3. All methods — comparison panel
| Method | Diversity | Subpart awareness | Interpretability score |
|---|---|---|---|
| PCA | Low | Low | baseline |
| RLDA | Medium | Medium | +0% |
| WDDWCC | Medium-High | Medium | +26% |
| **WDIWCD** | **High** | **High** | **+120%** |
| VAE+WDIWCD | High | Very High | best |


```python
methods = ["PCA", "RLDA", "WDDWCC", "WDIWCD", "VAE_WDIWCD"]
for m in methods:
    W = extract_weights(m, "mnist", n_components=16, seed=42)
    save_eigenimage_panel(W, f"figures/{m}_mnist_eigenimages.png",
                          title=f"{m} Eigenimages (MNIST)")
print("All eigenimage panels saved to interpretability/figures/")
```
